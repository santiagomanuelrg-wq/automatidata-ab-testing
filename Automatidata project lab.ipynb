### Task 3. Hypothesis testing

Before you conduct your hypothesis test, consider the following questions where applicable to complete your code response:

1. Recall the difference between the null hypothesis and the alternative hypotheses. Consider your hypotheses for this project as listed below.

$H_0$: There is no difference in the average fare amount between customers who use credit cards and customers who use cash.

$H_A$: There is a difference in the average fare amount between customers who use credit cards and customers who use cash.



Your goal in this step is to conduct a two-sample t-test. Recall the steps for conducting a hypothesis test: 


1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis 



**Note:** For the purpose of this exercise, your hypothesis test is the main component of your A/B test. 

You choose 5% as the significance level and proceed with a two-sample t-test.

```python
#==> ENTER YOUR CODE HERE
# --- Paso 0: Preparación de los datos ---
# Filtrar solo tarjeta de crédito (1) y efectivo (2)
taxi_filtered = taxi_data[taxi_data['payment_type'].isin([1, 2])].copy()

# Eliminar outliers de fare_amount (valores <= 0 y por encima del percentil 99)
q99 = taxi_filtered['fare_amount'].quantile(0.99)
taxi_filtered = taxi_filtered[
    (taxi_filtered['fare_amount'] > 0) &
    (taxi_filtered['fare_amount'] <= q99)
]

print("Dimensiones después del filtrado:", taxi_filtered.shape)
print("\nConteo por grupo:")
print(taxi_filtered['payment_type'].value_counts())
```

```python
# --- Paso 1: Plantear hipótesis nula (H0) y alternativa (HA) ---
# H0: No existe diferencia en el importe promedio de la tarifa entre
#     clientes que pagan con tarjeta de crédito y los que pagan en efectivo.
# HA: Existe una diferencia en el importe promedio de la tarifa entre
#     clientes que pagan con tarjeta de crédito y los que pagan en efectivo.

# --- Paso 2: Elegir el nivel de significancia ---
alpha = 0.05  # 5%

# --- Paso 3: Obtener el valor p ---
# Separar las muestras por grupo
credit_card = taxi_filtered[taxi_filtered['payment_type'] == 1]['fare_amount']
cash = taxi_filtered[taxi_filtered['payment_type'] == 2]['fare_amount']

print("\n=== Tamaño de cada muestra ===")
print(f"Tarjeta de crédito: n = {len(credit_card)}")
print(f"Efectivo:           n = {len(cash)}")

print("\n=== Medias y desviaciones estándar ===")
print(f"Tarjeta de crédito: media = {credit_card.mean():.4f}, std = {credit_card.std():.4f}")
print(f"Efectivo:           media = {cash.mean():.4f}, std = {cash.std():.4f}")

# --- Verificación de supuestos (recomendado) ---
# Prueba de Levene para homogeneidad de varianzas
levene_stat, levene_p = stats.levene(credit_card, cash)
print("\n=== Prueba de Levene (homogeneidad de varianzas) ===")
print(f"Estadístico = {levene_stat:.4f}, p-value = {levene_p:.4e}")

# Si p < 0.05 → varianzas desiguales → usar equal_var=False (t-test de Welch)
```

```python
# --- Prueba t para dos muestras independientes ---
t_stat, p_value = stats.ttest_ind(credit_card, cash, equal_var=False)

print("\n=== Resultado de la prueba t (Welch) ===")
print(f"Estadístico t = {t_stat:.4f}")
print(f"p-value       = {p_value:.4e}")

# --- Paso 4: Rechazar o no rechazar H0 ---
print("\n=== Decisión ===")
if p_value < alpha:
    print(f"Como p-value ({p_value:.4e}) < alpha ({alpha}), RECHAZAMOS H0.")
    print("Conclusión: Existe evidencia estadísticamente significativa de que")
    print("el importe promedio de la tarifa DIFIERE entre tarjeta y efectivo.")
else:
    print(f"Como p-value ({p_value:.4e}) >= alpha ({alpha}), NO rechazamos H0.")
    print("Conclusión: No hay evidencia suficiente para afirmar que el importe")
    print("promedio de la tarifa difiere entre tarjeta y efectivo.")
```

==> ENTER YOUR DECISION TO ACCEPT OR REJECT THE NULL HYPOTHESIS

The p-value is extremely small (much less than 0.05). This means that the probability of observing such a large difference (or greater) between the groups, if H₀ were true, is practically zero. Therefore, we reject the null hypothesis and conclude that there is a statistically significant difference in the average fare amount between customers who pay by credit card and those who pay in cash.
